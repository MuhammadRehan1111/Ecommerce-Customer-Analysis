```python
# Insights
# Insight 1 — Purchase Intent is the only significant driver of spending (p = 0.021).
# Among 12 categorical variables tested with one-way ANOVA,
#  only Purchase_Intent showed a statistically significant difference in purchase amount.
#  Wants-based customers spend the most ($292.01 avg), followed by Need-based ($280.80),
#  Planned ($270.34), and Impulsive ($256.84). This suggests emotional,
#  desire-driven purchases lead to higher spending than practical or spontaneous ones,
#   and no demographic factor (income, gender, education, device, payment method) influences how much a customer spends.

# Insight 2 — The spending pattern is consistent across income levels.
# The Income Level × Purchase Intent heatmap shows Wants-based customers spend
# the most in both income groups — $294 (High) and $290 (Middle) — while Impulsive buyers spend
# the least in both ($253 and $261). Income alone was not significant (p = 0.912).
# Emotion-driven buying transcends income: a Middle-income wants-based customer outspends a High-income impulsive buyer.

# Insight 3 — The customer base is remarkably homogeneous; no segment dominates.
# All score-based variables (Brand Loyalty, Product Rating, Customer Satisfaction, Frequency of Purchase, Time to Decision, Return Rate)
#  show near-uniform distributions, and demographic variables (Gender, Occupation, Payment Method, Device, Channel) are well balanced.
#   There is no single "typical heavy spender" segment — every customer type is equally represented in the data.


# Insight 4 — Purchase behaviour is clean, stable and outlier-free.
# Purchase amounts are uniformly distributed between ~$40 and $500 with median ≈ mean.
#The IQR method found zero outliers in all 9 numerical columns, and boxplots confirmed this visually.
#Monthly purchases remain steady throughout 2024 (63–100 per month) with only mild peaks in April and August.
#The business faces no seasonal risk and no abnormal transactions in this data.


# Insight 5 — Half the customer base is untouched by engagement programs.
# Discount usage is almost evenly split (52.1% used a discount, 47.9% did not) and loyalty membership is 49.1% members vs 50.9% non-members.
# This is a clear growth opportunity: roughly half of all customers have never used a discount or joined the loyalty program,
# and targeted campaigns could activate them.

# Insight 6 — Behavioural variables are largely independent of each other.
# Chi-square tests found no significant association between Income Level and Purchase Intent (p = 0.291) or between Device and
#Channel (p = 0.312), and the correlation heatmap showed no strong linear relationships among numerical variables.
#Each behaviour acts independently, so no single variable can be used as a proxy for another.

# Conclusion
# This project applied the complete EDA process to an e-commerce customer behaviour dataset of 1,000 records and 28 variables.
# The data was cleaned by converting currency-formatted amounts to numbers, fixing broken category labels,
# converting dates to proper datetime format, and filling missing values in two survey-style columns; no duplicate rows were found.
# Univariate analysis revealed a highly balanced, uniform customer base with no dominant segments and no outliers.
# Bivariate testing showed that purchase intent — not income, gender, device or any other factor — is the only significant predictor of spending,
# with wants-based buyers spending about 13% more than impulsive buyers, and multivariate analysis confirmed this pattern holds across income levels.

# Overall, the data is clean, stable and internally consistent.
# Future work could include:
# (1) clustering customers into segments for personalised marketing,especially the ~50% who are not loyalty members;
# (2) building a regression model to predict purchase amount from purchase intent and behavioural scores;
# (3) running A/B tests on discounts targeted at wants-based customers who already spend the most; and
# (4) validating these findings on real transactional data, since the near-uniform distributions suggest a synthetic or carefully balanced sample.
```
